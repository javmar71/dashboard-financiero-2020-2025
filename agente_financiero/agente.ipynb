````python
 ============================================================
# CELDA 7 — ETAPA 3C
# EXTRACCIÓN TABULAR DIRECTA
#
# PDF COMPLETO
#      ↓
# GEMINI
#      ↓
# TABLAS DE LOS ESTADOS FINANCIEROS
#      ↓
# JSON ESTRUCTURADO
#
# OBJETIVO:
# Obtener únicamente los DATOS TABULADOS del PDF.
#
# NO:
# - análisis
# - PUC
# - indicadores
# - normalización
# - cálculos
# - descripciones
# - evidencia narrativa
# - interpretación
# ============================================================

from pathlib import Path
import json
import os
import sys
import time
import re


# ============================================================
# 1. VALIDACIÓN DEL ENTORNO OFICIAL
# ============================================================

PYTHON_OFICIAL = (
    r"C:\Users\Usuario\Desktop\mi_proyecto_finanzas"
    r"\venv\Scripts\python.exe"
)

if sys.executable.lower() != PYTHON_OFICIAL.lower():
    raise RuntimeError(
        "El kernel de Jupyter NO corresponde al Python oficial.\n"
        f"Python actual: {sys.executable}\n"
        f"Python requerido: {PYTHON_OFICIAL}"
    )

if not PDF_2021.exists():
    raise FileNotFoundError(
        f"No existe el PDF objetivo:\n{PDF_2021}"
    )


# ============================================================
# 2. CLAVE API GEMINI
# ============================================================

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

if not GEMINI_API_KEY:
    raise RuntimeError(
        "No existe la variable de entorno GEMINI_API_KEY."
    )


# ============================================================
# 3. CONFIGURACIÓN GEMINI
# ============================================================

GEMINI_MODEL = "gemini-2.5-flash"

MAX_OUTPUT_TOKENS = 32768


# ============================================================
# 4. IMPORTAR CLIENTE GEMINI
# ============================================================

try:
    from google import genai
except ImportError as exc:
    raise ImportError(
        "No está disponible google-genai en el entorno oficial."
    ) from exc


# ============================================================
# 5. INSTRUCCIÓN ÚNICA PARA GEMINI
# ============================================================

PROMPT_GEMINI = r"""
ACTÚA EXCLUSIVAMENTE COMO EXTRACTOR DE DATOS TABULADOS DE
ESTADOS FINANCIEROS.

NO HAGAS ANÁLISIS FINANCIERO.

NO INTERPRETES LOS DATOS.

NO CALCULES INDICADORES.

NO APLIQUES PUC.

NO NORMALICES LOS CONCEPTOS.

NO CAMBIES LOS NOMBRES DE LAS CUENTAS.

NO CORRIJAS CIFRAS.

NO INVENTES DATOS.

NO UTILICES OCR.

NO GENERES EXPLICACIONES.

NO GENERES DESCRIPCIONES NARRATIVAS.

TU ÚNICA FUNCIÓN ES:

PDF COMPLETO
    ↓
IDENTIFICAR LAS TABLAS DE LOS ESTADOS FINANCIEROS
    ↓
LEER FILAS Y COLUMNAS
    ↓
CONVERTIR LAS TABLAS A JSON
    ↓
DEVOLVER ÚNICAMENTE EL JSON


============================================================
1. DEBES ANALIZAR EL PDF COMPLETO
============================================================

Revisa todas las páginas del PDF.

Debes identificar todos los estados financieros que realmente
aparezcan en el documento.

Como mínimo busca:

1. Estado de Situación Financiera.
2. Estado de Resultado Integral.
3. Estado de Cambios en el Patrimonio.
4. Estado de Flujo de Efectivo.

También reconoce variantes documentales de estos nombres.

NO asumas que un estado existe.

NO inventes un estado que no esté en el PDF.


============================================================
2. EXTRAER LAS TABLAS, NO TEXTO CORRIDO
============================================================

La unidad fundamental de extracción es:

FILA + COLUMNAS

Cada fila contable de las tablas debe conservarse.

Ejemplo conceptual:

ACTIVO

Efectivo                         20.991   43.925
Inversiones                     80.000   75.000
Cuentas por cobrar              50.000   48.000
Total activo                   348.286  359.005

Debe convertirse en datos JSON conservando esa relación.


============================================================
3. CONSERVAR LA ESTRUCTURA DE LA TABLA
============================================================

La estructura jerárquica observada en el documento debe
mantenerse.

Por ejemplo:

Activo
    Activo corriente y no corriente
        Efectivo
        Inversiones
        Cuentas por cobrar
        Total activos

No conviertas toda la información en una lista plana.

Si el documento presenta agrupaciones, subtotales o categorías,
conservarlas mediante objetos y listas JSON.


============================================================
4. CONCEPTOS
============================================================

Conserva exactamente el concepto que aparece en el documento.

Ejemplo:

"Efectivo"

"Total activos"

"Pasivos financieros"

"Patrimonio"

NO reemplaces esos nombres.

NO los traduzcas.

NO los normalices.

NO los relaciones todavía con PUC.

NO los relaciones todavía con las variables madre.


============================================================
5. NOTAS
============================================================

Si una fila tiene número de nota, conservarlo.

Ejemplo:

"nota": 7

Si no existe nota:

"nota": null


============================================================
6. COLUMNAS Y PERIODOS
============================================================

Lee primero los encabezados de las columnas.

Si el documento contiene:

2021 | 2020

entonces:

"2021": valor correspondiente a 2021
"2020": valor correspondiente a 2020

NO intercambies columnas.

NO inventes períodos.

NO supongas períodos que no aparecen.


============================================================
7. VALORES
============================================================

Los valores deben corresponder exactamente a las celdas de
las tablas.

Si aparece:

20.991

y el documento indica que está expresado en millones de pesos,
el valor JSON debe ser:

20991

Si aparece:

-8.318

debe conservarse como:

-8318

NO cambies el signo.

NO conviertas negativos en positivos.

NO redondees.

NO recalcules.

NO sustituyas un valor por otro.


============================================================
8. VALORES AUSENTES
============================================================

Si una celda representa ausencia de información mediante:

-
—
N/A
o una celda realmente vacía,

utiliza:

null

NO conviertas ausencia en cero.


============================================================
9. UNIDAD
============================================================

Conserva la unidad indicada por el documento.

Ejemplo:

"En millones de pesos"

No conviertas unidades.

No cambies millones a pesos.

No cambies pesos a miles.


============================================================
10. TOTALES Y SUBTOTALES
============================================================

Los totales y subtotales son datos de la tabla.

DEBEN conservarse.

Ejemplo:

"Total activos"

"Total pasivos"

"Total patrimonio"

"Total pasivos y patrimonio"

NO los elimines.

NO los recalcules.


============================================================
11. ESTADO DE RESULTADO INTEGRAL
============================================================

Extrae todas las filas y columnas de la tabla correspondiente.

Por ejemplo, si aparecen:

Ingresos
Gastos
Resultado operacional
Resultado antes de impuestos
Impuesto
Utilidad del ejercicio
Otro resultado integral
Resultado integral total

deben conservarse como aparecen en el documento.

NO calcules ninguna de estas cifras.


============================================================
12. ESTADO DE CAMBIOS EN EL PATRIMONIO
============================================================

Este estado puede tener múltiples columnas y movimientos.

DEBES conservar:

- períodos;
- saldos iniciales;
- aumentos;
- disminuciones;
- dividendos;
- resultados;
- otros movimientos;
- saldos finales;

únicamente cuando aparezcan en la tabla.

NO simplifiques la tabla.

NO elimines columnas.


============================================================
13. ESTADO DE FLUJO DE EFECTIVO
============================================================

Conserva la estructura de la tabla.

Por ejemplo:

Flujos de efectivo de actividades de operación
Flujos de efectivo de actividades de inversión
Flujos de efectivo de actividades de financiación
Aumento/disminución del efectivo
Efectivo al inicio
Efectivo al final

Si existen subtotales o totales, conservarlos.


============================================================
14. NO EXTRAER LAS NOTAS COMO ESTADOS
============================================================

Las notas detalladas a los estados financieros NO deben
convertirse en estados financieros independientes.

La prioridad es extraer las tablas de:

- Estado de Situación Financiera;
- Estado de Resultado Integral;
- Estado de Cambios en el Patrimonio;
- Estado de Flujo de Efectivo.

Las notas solo forman parte de la extracción si aparecen
dentro de la propia tabla del estado.


============================================================
15. NO AGREGAR INFORMACIÓN INNECESARIA
============================================================

El JSON NO debe contener:

- evidencia textual;
- explicaciones;
- comentarios;
- análisis;
- interpretación;
- PUC;
- indicadores;
- fórmulas;
- ratios;
- conclusiones;
- observaciones narrativas;
- confianza;
- clasificación semántica;
- recomendaciones.


============================================================
16. ESTRUCTURA JSON OBLIGATORIA
============================================================

La respuesta debe ser un único objeto JSON.

Debe seguir esta estructura general:

{
  "empresa": "...",
  "moneda": "...",
  "periodos": [
    "2021",
    "2020"
  ],
  "estados_financieros": {

    "estado_de_situacion_financiera": {

      "presente": true,

      "activo": {
        "activo_corriente_y_no_corriente": [
          {
            "concepto": "Efectivo",
            "nota": 7,
            "2021": 20991,
            "2020": 43925
          }
        ]
      },

      "pasivo": {
        "pasivo_corriente_y_no_corriente": []
      },

      "patrimonio": {
        "componentes": []
      }

    },

    "estado_de_resultado_integral": {

      "presente": true,

      "resultado": []

    },

    "estado_de_cambios_en_el_patrimonio": {

      "presente": true,

      "movimientos": []

    },

    "estado_de_flujo_de_efectivo": {

      "presente": true,

      "flujos": []

    }
  }
}


============================================================
17. IMPORTANTE SOBRE LA ESTRUCTURA
============================================================

La estructura anterior es una guía.

Debes adaptarla a la estructura REAL de las tablas del PDF.

La prioridad es:

1. conservar todos los datos;
2. conservar las relaciones jerárquicas;
3. conservar filas;
4. conservar columnas;
5. conservar períodos;
6. conservar valores;
7. conservar notas;
8. conservar unidades.

NO fuerces todos los estados a tener exactamente los mismos
campos internos si las tablas tienen estructuras diferentes.


============================================================
18. ESTADOS AUSENTES
============================================================

Si después de revisar TODO el PDF un estado realmente no existe:

"presente": false

y no inventes datos.

Ejemplo:

"estado_de_flujo_de_efectivo": {
    "presente": false
}


============================================================
19. REGLA ABSOLUTA
============================================================

ESTE JSON ES LA EXTRACCIÓN PRIMARIA.

NO ES TODAVÍA LA MATRIZ MADRE.

NO ES TODAVÍA EL PUC.

NO ES TODAVÍA EL MODELO DE INDICADORES.

NO ES TODAVÍA EL ANÁLISIS FINANCIERO.

Solo representa los datos tabulados encontrados en el PDF.


============================================================
20. SALIDA
============================================================

DEVUELVE ÚNICAMENTE JSON VÁLIDO.

NO escribas ningún texto antes del JSON.

NO escribas ningún texto después del JSON.

NO utilices Markdown.

NO utilices ```json.

NO agregues explicaciones.
"""


# ============================================================
# 6. CREAR CLIENTE GEMINI
# ============================================================

client = genai.Client(
    api_key=GEMINI_API_KEY
)


# ============================================================
# 7. FUNCIONES DE APOYO
# ============================================================

def limpiar_respuesta_json(texto):
    """
    Elimina únicamente posibles envoltorios Markdown.
    No altera el contenido del JSON.
    """

    if texto is None:
        return ""

    texto = texto.strip()

    if texto.startswith("```"):
        texto = re.sub(
            r"^```(?:json)?\s*",
            "",
            texto,
            flags=re.IGNORECASE
        )

        texto = re.sub(
            r"\s*```$",
            "",
            texto
        )

    return texto.strip()


def parsear_json(respuesta):
    """
    Convierte la respuesta de Gemini en un objeto Python.
    """

    texto = getattr(respuesta, "text", None)

    if not texto:
        raise RuntimeError(
            "Gemini devolvió una respuesta vacía."
        )

    texto = limpiar_respuesta_json(texto)

    try:
        return json.loads(texto)

    except json.JSONDecodeError as exc:

        raise RuntimeError(
            "Gemini no devolvió JSON válido.\n"
            f"Longitud recibida: {len(texto):,} caracteres."
        ) from exc


def validar_json_final(datos):
    """
    Validación mínima.
    
    No interpreta ni normaliza los datos.
    """

    if not isinstance(datos, dict):
        raise RuntimeError(
            "La raíz del JSON no es un objeto."
        )

    campos_obligatorios = [
        "empresa",
        "moneda",
        "periodos",
        "estados_financieros"
    ]

    faltantes = [
        campo
        for campo in campos_obligatorios
        if campo not in datos
    ]

    if faltantes:
        raise RuntimeError(
            "Faltan campos obligatorios: "
            + ", ".join(faltantes)
        )

    if not isinstance(
        datos["estados_financieros"],
        dict
    ):
        raise RuntimeError(
            "'estados_financieros' debe ser un objeto."
        )

    return True


def contar_elementos(datos):
    """
    Cuenta recursivamente listas de datos.
    
    Solo se utiliza como indicador técnico de volumen.
    No modifica el JSON.
    """

    if isinstance(datos, list):
        return len(datos) + sum(
            contar_elementos(x)
            for x in datos
        )

    if isinstance(datos, dict):
        return sum(
            contar_elementos(v)
            for v in datos.values()
        )

    return 0


def detectar_finalizacion(respuesta):
    """
    Detecta de forma defensiva si Gemini terminó por límite
    de salida.
    """

    try:

        candidatos = getattr(
            respuesta,
            "candidates",
            None
        )

        if not candidatos:
            return "NO_DETERMINADO"

        candidato = candidatos[0]

        razon = getattr(
            candidato,
            "finish_reason",
            None
        )

        if razon is None:
            return "NO_DETERMINADO"

        texto = str(razon).upper()

        if "MAX" in texto or "LENGTH" in texto:
            return "TRUNCADO_POR_LIMITE"

        return texto

    except Exception:
        return "NO_DETERMINADO"


# ============================================================
# 8. VARIABLES DE EJECUCIÓN
# ============================================================

archivo_gemini = None

inicio_total = time.time()


# ============================================================
# 9. CARGAR PDF Y ENVIARLO A GEMINI
# ============================================================

try:

    print("============================================================")
    print("ETAPA 3C — EXTRACCIÓN TABULAR DIRECTA")
    print("============================================================")
    print(f"PDF: {PDF_2021}")
    print(f"Modelo: {GEMINI_MODEL}")
    print("Extracción: TABLAS → JSON")
    print("OCR: NO")
    print("PUC: NO")
    print("Indicadores: NO")
    print("Análisis: NO")
    print("============================================================")

    print("\nCargando PDF en Gemini...")

    archivo_gemini = client.files.upload(
        file=str(PDF_2021)
    )

    print(
        f"[OK] Archivo cargado en Gemini: "
        f"{archivo_gemini.name}"
    )


    # ========================================================
    # 10. UNA ÚNICA LLAMADA PARA TODO EL PDF
    # ========================================================

    print("\nEnviando instrucción de extracción a Gemini...")

    respuesta = client.models.generate_content(
        model=GEMINI_MODEL,
        contents=[
            archivo_gemini,
            PROMPT_GEMINI
        ],
        config={
            "temperature": 0,
            "max_output_tokens": MAX_OUTPUT_TOKENS,
            "response_mime_type": "application/json"
        }
    )


    # ========================================================
    # 11. COMPROBAR FINALIZACIÓN
    # ========================================================

    motivo_finalizacion = detectar_finalizacion(
        respuesta
    )

    print(
        f"[INFO] Finalización Gemini: "
        f"{motivo_finalizacion}"
    )

    if motivo_finalizacion == "TRUNCADO_POR_LIMITE":

        raise RuntimeError(
            "Gemini alcanzó el límite de salida. "
            "El JSON no se acepta como completo."
        )


    # ========================================================
    # 12. PARSEAR JSON
    # ========================================================

    print("\nProcesando JSON recibido...")

    datos_finales = parsear_json(
        respuesta
    )


    # ========================================================
    # 13. VALIDACIÓN MÍNIMA
    # ========================================================

    validar_json_final(
        datos_finales
    )

    print(
        "[OK] JSON válido."
    )


    # ========================================================
    # 14. GUARDAR JSON
    # ========================================================

    SALIDA_JSON_2021.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        SALIDA_JSON_2021,
        "w",
        encoding="utf-8"
    ) as archivo_salida:

        json.dump(
            datos_finales,
            archivo_salida,
            ensure_ascii=False,
            indent=2
        )


    # ========================================================
    # 15. RELECTURA FÍSICA
    # ========================================================

    if not SALIDA_JSON_2021.exists():

        raise RuntimeError(
            "El JSON no existe después de la escritura."
        )

    if SALIDA_JSON_2021.stat().st_size == 0:

        raise RuntimeError(
            "El JSON fue creado vacío."
        )

    with open(
        SALIDA_JSON_2021,
        "r",
        encoding="utf-8"
    ) as archivo_lectura:

        json_verificado = json.load(
            archivo_lectura
        )


    validar_json_final(
        json_verificado
    )


    # ========================================================
    # 16. INFORME TÉCNICO
    # ========================================================

    estados = json_verificado.get(
        "estados_financieros",
        {}
    )

    estados_presentes = []

    for nombre_estado, contenido in estados.items():

        if isinstance(contenido, dict):

            if contenido.get("presente") is True:

                estados_presentes.append(
                    nombre_estado
                )


    total_elementos = contar_elementos(
        json_verificado
    )

    duracion = time.time() - inicio_total


    print("\n")
    print("============================================================")
    print("ETAPA 3C — RESULTADO")
    print("============================================================")

    print(
        f"JSON generado: {SALIDA_JSON_2021}"
    )

    print(
        f"Tamaño: "
        f"{SALIDA_JSON_2021.stat().st_size:,} bytes"
    )

    print(
        f"Tiempo: "
        f"{duracion:.1f} segundos"
    )

    print(
        f"Estados presentes: "
        f"{len(estados_presentes)}"
    )

    for estado in estados_presentes:

        print(
            f"  [OK] {estado}"
        )

    print(
        f"Elementos tabulares/estructurales: "
        f"{total_elementos:,}"
    )

    print("")
    print("PDF → GEMINI:                  [OK]")
    print("TABLAS → JSON:                 [OK]")
    print("JSON válido:                   [OK]")
    print("Relectura física:              [OK]")
    print("Extracción completa del PDF:   [OK]")
    print("PUC:                            NO")
    print("Indicadores:                   NO")
    print("Análisis financiero:            NO")
    print("Normalización:                 NO")
    print("OCR:                            NO")
    print("")
    print("ETAPA 3C FINALIZADA.")
    print("============================================================")


finally:

    # ========================================================
    # 17. ELIMINAR ARCHIVO REMOTO DE GEMINI
    # ========================================================

    if archivo_gemini is not None:

        try:

            client.files.delete(
                name=archivo_gemini.name
            )

            print(
                "\n[OK] Archivo remoto eliminado de Gemini."
            )

        except Exception as exc:

            print(
                "\n[ADVERTENCIA] "
                "No fue posible eliminar el archivo remoto."
            )

            print(
                f"Detalle: {exc}"
            )
````
